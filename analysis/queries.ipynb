# Extract the Query Information


```python
import csv
import os
import simplejson as json

from log import log
from util import smart_open, sort_query_list


def extract_query_info(dataset: str = "stackoverflow", version: str = "v1.1"):
    os.makedirs(f"queries/{version}", exist_ok=True)
    queries_dir = f"../{version}/{dataset}/queries_generated"
    output_file = f"queries/{version}/{dataset}.csv.gz"

    log.info(f"Extracting query information for {dataset} version {version}")
    data = {}
    keys = []
    for file in os.listdir(queries_dir):
        if not file.endswith(".sql"):
            continue

        with open(os.path.join(queries_dir, file), 'r') as query_file:
            lines = query_file.read().splitlines()
            info = json.loads(lines[0].replace("--", "").strip())
            lines = lines[1:]

        row = {
            "query": file,
            "model": info["model"],
            "prompt": info["prompt"],
            "temperature": info["temperature"],
            "reasoning": info["reasoning"],
            "input_tokens": info["input_tokens"],
            "output_tokens": info["output_tokens"],
            "bytes": len("\n".join(lines).encode('utf-8')),
            "characters": len("\n".join([l.strip() for l in lines])),
            "words": len(" ".join([l.strip() for l in lines]).split()),
            "lines": len(lines),
        }

        data[file] = row
        if not keys:
            keys = list(row.keys())

    with smart_open(output_file, 'wt') as out_file:
        fieldnames = keys
        writer = csv.DictWriter(out_file, fieldnames=fieldnames)
        writer.writeheader()
        for query in sort_query_list(data.keys()):
            writer.writerow(data[query])


extract_query_info(dataset="stackoverflow", version="v1.1")
extract_query_info(dataset="stackoverflow", version="v1.2")
extract_query_info(dataset="stackoverflow", version="v1.3")
extract_query_info(dataset="stackoverflow", version="v1.4")
extract_query_info(dataset="stackoverflow", version="v1.5")
extract_query_info(dataset="stackoverflow", version="v2.0")
```
