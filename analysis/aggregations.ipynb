# Extract the Query Plans


```python
import os
import queryplans
from log import log


def extract_queryplans(dataset: str = "stackoverflow_dba", version: str = "v1.0", sqlstorm: bool = True):
    os.makedirs(f"queryplans/{version}", exist_ok=True)
    input_file = f"../results/{version}/features/{dataset}{'_sqlstorm_' + version if sqlstorm else ''}.csv"
    output_file = f"queryplans/{version}/{dataset}{'_sqlstorm' if sqlstorm else ''}.csv.gz"

    log.info(f"Extracting queryplans for {dataset} version {version} (SQLStorm: {sqlstorm})")
    with log.file(output_file.replace(".csv.gz", ".log")) as log_file:
        queryplans.compute(input_file, output_file=output_file)
```

```python
extract_queryplans(dataset="job", version="v0.0", sqlstorm=False)
extract_queryplans(dataset="tpchSf1", version="v0.0", sqlstorm=True)
extract_queryplans(dataset="tpcdsSf1", version="v0.0", sqlstorm=True)

extract_queryplans(dataset="stackoverflow_dba", version="v1.0", sqlstorm=True)
extract_queryplans(dataset="job", version="v1.0", sqlstorm=True)
extract_queryplans(dataset="tpchSf1", version="v1.0", sqlstorm=True)
extract_queryplans(dataset="tpcdsSf1", version="v1.0", sqlstorm=True)
```

```python
extract_queryplans(dataset="stackoverflow_dba", version="v2.0", sqlstorm=True)
```

# Compute the Aggregations


```python
import os
import queryplans
from log import log


def extract_aggregations(dataset: str = "stackoverflow_dba", version: str = "v1.0", sqlstorm: bool = True):
    os.makedirs(f"aggregations/{version}", exist_ok=True)
    input_file = f"../results/{version}/features/{dataset}{'_sqlstorm_' + version if sqlstorm else ''}.csv"
    output_file = f"aggregations/{version}/{dataset}{'_sqlstorm' if sqlstorm else ''}.csv.gz"

    log.info(f"Extracting aggregations for {dataset} version {version} (SQLStorm: {sqlstorm})")
    with log.file(output_file.replace(".csv.gz", ".log")) as log_file:
        queryplans.compute_aggregations(input_file, output_file=output_file)
```

```python
extract_aggregations(dataset="job", version="v0.0", sqlstorm=False)
extract_aggregations(dataset="tpchSf1", version="v0.0", sqlstorm=True)
extract_aggregations(dataset="tpcdsSf1", version="v0.0", sqlstorm=True)

extract_aggregations(dataset="stackoverflow_dba", version="v1.0", sqlstorm=True)
extract_aggregations(dataset="job", version="v1.0", sqlstorm=True)
extract_aggregations(dataset="tpchSf1", version="v1.0", sqlstorm=True)
extract_aggregations(dataset="tpcdsSf1", version="v1.0", sqlstorm=True)

extract_aggregations(dataset="stackoverflow_dba", version="v2.0", sqlstorm=True)
```
